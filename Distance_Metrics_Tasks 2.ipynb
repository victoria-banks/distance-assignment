## Distance Metrics: Manhattan, Euclidean, and Minkowski

This notebook builds on the *Euclidean and Manhattan Distance Calculations* case study (same dataset, same style) to compute distances from every point to the reference point (6, 6[, 6]) using several distance metrics, and to compare their distributions.

```python
import numpy as np
import pandas as pd
import os
import matplotlib.pyplot as plt
import matplotlib.cm as cm
%matplotlib inline
os.makedirs('figures', exist_ok=True)

plt.style.use('ggplot')
```

### Load the dataset

```python
df = pd.read_csv('distance_dataset.csv', index_col=0)
df.head()
```

### Task 1: Manhattan distance from each (X, Y) to (6, 6)
Sum of absolute differences along each axis to the reference point.

```python
manhattan_xy_to_66 = np.abs(df.X - 6) + np.abs(df.Y - 6)
manhattan_xy_to_66.values
```

### Task 2: Euclidean distance from each (X, Y) to (6, 6)
Straight-line distance to the reference point.

```python
euclidean_xy_to_66 = np.sqrt((df.X - 6)**2 + (df.Y - 6)**2)
euclidean_xy_to_66.values
```

### Task 3: Minkowski distance (p = 1.5) from each (X, Y) to (6, 6)
The Minkowski distance generalizes both Manhattan (p=1) and Euclidean (p=2) distance:

$$ D = \left( |x_1 - x_2|^p + |y_1 - y_2|^p \right)^{1/p} $$

With `p = 1.5` we get a distance that sits between Manhattan and Euclidean.

```python
p = 1.5
minkowski_xy_to_66 = (np.abs(df.X - 6)**p + np.abs(df.Y - 6)**p) ** (1 / p)
minkowski_xy_to_66.values
```

### Task 4: Minkowski distance (p = 1.5) from each (X, Y, Z) to (6, 6, 6)
Same formula extended to three dimensions.

```python
minkowski_xyz_to_666 = (np.abs(df.X - 6)**p + np.abs(df.Y - 6)**p + np.abs(df.Z - 6)**p) ** (1 / p)
minkowski_xyz_to_666.values
```

*(Sanity check: these can also be computed with `scipy.spatial.distance` functions row by row, or with `np.linalg.norm(..., ord=p, axis=1)` — both should match the arrays above.)*

```python
# Sanity check using scipy / numpy built-ins
import scipy.spatial.distance as dist

check_manhattan = np.array([dist.cityblock([x, y], [6, 6]) for x, y in zip(df.X, df.Y)])
check_euclidean = np.array([dist.euclidean([x, y], [6, 6]) for x, y in zip(df.X, df.Y)])
check_mink_xy = np.linalg.norm(df[['X', 'Y']].values - [6, 6], ord=p, axis=1)
check_mink_xyz = np.linalg.norm(df[['X', 'Y', 'Z']].values - [6, 6, 6], ord=p, axis=1)

print('Manhattan matches:', np.allclose(manhattan_xy_to_66, check_manhattan))
print('Euclidean matches:', np.allclose(euclidean_xy_to_66, check_euclidean))
print('Minkowski (x,y) matches:', np.allclose(minkowski_xy_to_66, check_mink_xy))
print('Minkowski (x,y,z) matches:', np.allclose(minkowski_xyz_to_666, check_mink_xyz))
```

### Task 5 (bonus / graduate): Histograms comparing all four distance distributions

```python
fig, axs = plt.subplots(1, 4, figsize=(24, 5), sharey=True)

axs[0].hist(manhattan_xy_to_66, bins=30, color='red')
axs[0].set_title('Manhattan (X,Y) to (6,6)')

axs[1].hist(euclidean_xy_to_66, bins=30, color='blue')
axs[1].set_title('Euclidean (X,Y) to (6,6)')

axs[2].hist(minkowski_xy_to_66, bins=30, color='green')
axs[2].set_title('Minkowski p=1.5 (X,Y) to (6,6)')

axs[3].hist(minkowski_xyz_to_666, bins=30, color='purple')
axs[3].set_title('Minkowski p=1.5 (X,Y,Z) to (6,6,6)')

for ax in axs:
    ax.set_xlabel('Distance')
axs[0].set_ylabel('Count')

fig.suptitle('Comparison of Distance Metric Distributions (reference point at 6, 6[, 6])', fontsize=16)
plt.tight_layout()
plt.savefig('figures/TaskDistanceComparison.png')
plt.show()
```

**Observation:** the Manhattan distances are shifted right relative to Euclidean, since summing absolute axis-wise differences is always ≥ the straight-line distance. The Minkowski (p=1.5) distribution sits between the two, as expected since p=1 is Manhattan and p=2 is Euclidean. Adding the Z dimension shifts the 3D Minkowski distances further right, since there's a third axis contributing additional distance.
